## Generar lexicon aranes

````python
import re
import pdfplumber

PDF_PATH = "../../diccionarioAranesí.pdf"
# OUTPUT_PATH = "lexicons/aranes.txt"

def es_negrita(span):
    font = span.get("fontname", "").lower()
    return "bold" in font or "bd" in font

def limpiar_palabra(p):
    # Elimina caracteres no válidos y números
    p = re.sub(r"[^\wáéíóúüñç·'-]", "", p.lower())
    p = re.sub(r"\d", "", p)
    return p

def expandir_guiones(palabra):
    """
    Maneja casos como:
    - "-a" → "" (se ignora)
    - "abreujador" → ["abreujador"]
    - "abreujador-a" → ["abreujador", "abreujadora"]
    - "abans-arrèhilh" → ["abans-arrèhilh"] (palabra compuesta real)
    """
    resultados = set()

    # Quitar guion inicial
    if palabra.startswith("-"):
        # palabra = palabra[1:]
        return []

    # Si no hay guion, devolver tal cual
    if "-" not in palabra:
        if palabra:
            resultados.add(palabra)
        return resultados

    # Si hay un guion en medio, puede ser flexión tipo "-a"
    partes = palabra.split("-")
    # Caso típico: raíz + sufijo flexivo
    if len(partes[-1]) <= 4:
        sufijo = partes[-1]
        raiz = "-".join(partes[:-1])
        if raiz:
            resultados.add(raiz)
            resultados.add(raiz + sufijo)
        return resultados

    # Si es palabra compuesta real (ej. abans-arrèhilh)
    resultados.add(palabra)
    return resultados

def extraer_entradas(pdf_path, skip_pages=0):
    entradas = set()

    with pdfplumber.open(pdf_path) as pdf:
        for i, page in enumerate(pdf.pages):
            if i < skip_pages:
                continue

            spans = page.extract_words(extra_attrs=["fontname"])

            for sp in spans:
                if es_negrita(sp):
                    palabra = sp["text"].strip().split(" ")[0]
                    palabra = limpiar_palabra(palabra)

                    if not palabra:
                        continue

                    for forma in expandir_guiones(palabra):
                        if forma and not re.search(r"\d", forma):
                            entradas.add(forma)

    return entradas

def guardar_txt(entradas, output_path):
    with open(output_path, "w", encoding="utf-8") as f:
        for palabra in sorted(entradas):
            f.write(palabra + "\n")

if __name__ == "__main__":
    n = 10
    entradas = extraer_entradas(PDF_PATH, skip_pages=n)
    guardar_txt(entradas, OUTPUT_PATH)
    print(f"Generado {OUTPUT_PATH} con {len(entradas)} palabras (saltadas {n} páginas).")
````

### Vocabulario

````python
from generateDataset import generateDatasetHuecos
huecosDataset = generateDatasetHuecos(ast, False)
print(huecosDataset.shape)
````

````python
from lowresource_llm_evaluation.vocabulario import evaluacionHuecos
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

# Nombre del modelo en Hugging Face
model_name = "google/gemma-3-1b-it"

# 1. Cargar tokenizador
tokenizer = AutoTokenizer.from_pretrained(model_name)

# 2. Cargar modelo
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    dtype=torch.float32,
    device_map="auto"   # opcional: usa GPU si está disponible
)
acc_vals = []
acc_low_vals = []
lev_vals = []
for _, row in huecosDataset.iterrows():
    res = evaluacionHuecos(
        model, tokenizer,
        row["masked_sentence"],
        row["missing_word"],
        lang="ast"
    )
    acc_vals.append(res["accuracy"])
    acc_low_vals.append(res["accuracy_lower"])
    lev_vals.append(res["levenshtein"])
    print(res)
print(float(np.mean(acc_vals)))
print(float(np.mean(acc_low_vals)))
print(float(np.mean(lev_vals)))
````
