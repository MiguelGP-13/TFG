## Modelo en zip

```python
import os
import zipfile

def zip_and_link(folder_path, zip_name):
    # Crear ZIP
    with zipfile.ZipFile(zip_name, 'w', zipfile.ZIP_DEFLATED) as z:
        for root, dirs, files in os.walk(folder_path):
            for file in files:
                full_path = os.path.join(root, file)
                rel_path = os.path.relpath(full_path, folder_path)
                z.write(full_path, rel_path)

zip_and_link("/notebooks/qlora_gallego__Qwen2.5-7B-Instruct_05-21_06-53-27/checkpoint-15000","Gallego15000.zip")
```

## Abrir zip

```python
import os
import zipfile

def unpack_zip(zip_path, root=None):
    # Nombre base del zip sin extensión
    base_name = os.path.splitext(os.path.basename(zip_path))[0]

    # Si root no se especifica, crea carpeta con el nombre del zip
    if root is None:
        extract_dir = base_name
    else:
        extract_dir = root

    # Crear carpeta si no existe
    os.makedirs(extract_dir, exist_ok=True)

    # Extraer contenido
    with zipfile.ZipFile(zip_path, 'r') as z:
        z.extractall(extract_dir)

    return extract_dir
unpack_zip("/notebooks/modelo3200SinConcatenar.zip")
```

## Recuperar archivo de la papelera

```python
import os
import shutil


def find_and_copy_checkpoint(root, target, destination="."):
    for dirpath, dirnames, filenames in os.walk(root):
        for d in dirnames:
            if target in d:
                src = os.path.join(dirpath, d)
                dst = os.path.join(destination, d)
                print(f"Copiando {src} → {dst}")
                shutil.copytree(src, dst)
                return dst  # ← aquí hace return inmediatamente
    print("No se encontró ninguna carpeta con ese nombre.")
    return None

find_and_copy_checkpoint(".Trash-0/files", "qlora_aranes__Qwen2.5-7B-Instruct_05-24_14-33-15")
```

## Ver tamaño de carpeta

```python
import os

def obtener_tamano_carpeta(ruta):
    total_size = 0
    # os.walk recorre la carpeta, subcarpetas y archivos
    for dirpath, dirnames, filenames in os.walk(ruta):
        for f in filenames:
            fp = os.path.join(dirpath, f)
            # sumamos si el archivo existe y no es un enlace roto
            if os.path.exists(fp):
                total_size += os.path.getsize(fp)
    
    # Convertir a formato legible (MB, GB)
    for unidad in ['B', 'KB', 'MB', 'GB', 'TB']:
        if total_size < 1024:
            return f"{total_size:.2f} {unidad}"
        total_size /= 1024

print(obtener_tamano_carpeta('.Trash-0'))
```

## Resultados en zip

```python
import os
import glob
import zipfile
import time

def comprimir_y_eliminar_json_time(ruta_carpeta):
    # Generar la fecha de hoy usando la librería time
    # %Y = Año, %m = Mes, %d = Día
    fecha_hoy = time.strftime("%Y-%m-%d")
    nombre_zip = f"resultados_{fecha_hoy}.zip"
    
    # Patrón de búsqueda: resultados*.json
    patron = os.path.join(ruta_carpeta, "resultados*.json")
    archivos_encontrados = glob.glob(patron)

    if not archivos_encontrados:
        print("No se encontraron archivos que coincidan con 'resultados*.json'.")
        return

    ruta_zip = os.path.join(ruta_carpeta, nombre_zip)

    try:
        # 1. Crear el ZIP y meter los archivos
        with zipfile.ZipFile(ruta_zip, 'w', zipfile.ZIP_DEFLATED) as zipf:
            for archivo in archivos_encontrados:
                nombre_base = os.path.basename(archivo)
                zipf.write(archivo, nombre_base)
                print(f"Empaquetado: {nombre_base}")

        print(f"\nÉxito: Archivo '{nombre_zip}' creado.")

        # 2. Limpieza de los archivos sueltos
        for archivo in archivos_encontrados:
            os.remove(archivo)
            print(f"Borrado del disco: {os.path.basename(archivo)}")
            
        print("\nTarea completada.")

    except Exception as e:
        print(f"Error durante el proceso: {e}")

# Ejemplo de ejecución:
comprimir_y_eliminar_json_time('.')
```
